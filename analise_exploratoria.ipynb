# Análise Explorotária dos dados

**Análise exploratória das base de dados do ENEM 2025**

Disponível em: [Microdados ENEM](https://www.gov.br/inep/pt-br/acesso-a-informacao/dados-abertos/microdados/enem)

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
```

```python
# Leitura dos dados
dados = np.genfromtxt('microdados_enem_2025/DADOS/RESULTADOS_2025.csv', 
                      delimiter=';', 
                      skip_header=1, 
                      encoding='latin-1', 
                      usecols=(12, 22, 23, 24, 25, 30, 36, 37, 38, 39, 40, 41))

# Colunas selecionadas:
# UF, NOTA_CN, NOTA_CH, NOTA_LC, NOTA_MT, LINGUA, NOTA_R1, NOTA_R2, NOTA_R3, NOTA_R4, NOTA_R5, NOTA_RT
```

```python
linhas_com_nan = np.any(np.isnan(dados), axis=1)

linhas_validas = ~linhas_com_nan

dados_limpos = dados[linhas_validas]

print(f"Número de linhas originais: {dados.shape[0]}")
print(f"Número de linhas da base limpa: {dados_limpos.shape[0]}")
```

```python
cn = dados_limpos[:,1]
ch = dados_limpos[:,2]

# Histograma 2D
plt.figure(figsize=(8, 6))
plt.hist2d(cn, ch, bins=500, cmap='viridis', norm=mcolors.LogNorm())

plt.colorbar(label='Frequência')
plt.xlabel('Nota em Ciências da Natureza')
plt.ylabel('Nota em Ciências Humanas')
plt.title('Relação entre a nota de Ciências da Natura e Ciências Humanas')

plt.show()
```

```python
cn = dados_limpos[:,1]
ch = dados_limpos[:,2]

# Retira as notas 0
positivos = (cn>0) & (ch>0)

cn = cn[positivos]
ch = ch[positivos]

# Histograma 2D
plt.figure(figsize=(8, 6))
plt.hist2d(cn, ch, bins=500, cmap='viridis', norm=mcolors.LogNorm())

plt.colorbar(label='Frequência')
plt.xlabel('Nota em Ciências da Natureza')
plt.ylabel('Nota em Ciências Humanas')
plt.title('Relação entre a nota de Ciências da Natura e Ciências Humanas')

plt.show()
```
